```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

import sys

# ============================================================
# 1. Add project root to Python path
# ============================================================

sys.path.append("/Workspace/spotifyazureproject/spotify_lab")

from utils.transformation import reusable


# ============================================================
# 2. Read new data from Bronze using Auto Loader
# ============================================================

df_user = (
    spark.readStream
        .format("cloudFiles")
        .option("cloudFiles.format", "parquet")
        .option(
            "cloudFiles.schemaLocation",
            "abfss://silver@storagemikeproject.dfs.core.windows.net/DimUser/schema"
        )
        .load(
            "abfss://bronze@storagemikeproject.dfs.core.windows.net/DimUser"
        )
)


# ============================================================
# 3. Silver transformations
# ============================================================

df_user_obj = reusable()

# Remove technical column
df_user = df_user_obj.dropColumns(
    df_user,
    ["_rescued_data"]
)

# Remove duplicate users
df_user = df_user.dropDuplicates(
    ["user_id"]
)


# ============================================================
# 4. Write data to Silver Delta
# ============================================================

query = (
    df_user.writeStream
        .format("delta")
        .outputMode("append")
        .option(
            "checkpointLocation",
            "abfss://silver@storagemikeproject.dfs.core.windows.net/DimUser/checkpoint"
        )
        .trigger(availableNow=True)
        .start(
            "abfss://silver@storagemikeproject.dfs.core.windows.net/DimUser/data"
        )
)

query.awaitTermination()

print("DimUser Silver processing completed successfully.")
```
